```python
import subprocess

sentence = 'Two boys are sitting on a couch that is red and black.'
checkpoint = '../part4/nmt.model'
config = 'model_config.yaml'

top_k_vals = [3, 7, 10, 25, 50]
top_p_vals = [0.3, 0.5, 0.7, 0.8, 0.9, 1.0]
temps      = [0.3, 0.5, 0.7, 0.9, 1.0]

results = []

cmd = (f"python predict.py --sentence '{sentence}' --checkpoint-path {checkpoint} --model-config {config}")
result = subprocess.run(cmd, shell=True, capture_output=True, text=True) # https://stackoverflow.com/questions/74763554/how-to-use-subprocess-run-method-in-python
print("No sampling:\n", result.stdout) # https://www.reddit.com/r/learnpython/comments/7tolo2/capturing_output_from_subprocessrun/
results.append(f"No sampling:\n{result.stdout}\n")

for k in top_k_vals:
    for p in top_p_vals:
        for t in temps:
            print(f"\ncombo: top-k={k}, top-p={p}, temp={t}")
            cmd = (f"python predict.py --do-sample  --sentence '{sentence}' --checkpoint-path {checkpoint} --model-config {config} --top-k {k} --top-p {p} --temperature {t}")
            result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
            print(result.stdout)
            results.append(f"top-k={k}, top-p={p}, temp={t}:\n{result.stdout}\n")
```

```python
with open('part4_results.txt', 'w') as f:
    f.write('\n'.join(results))
```
